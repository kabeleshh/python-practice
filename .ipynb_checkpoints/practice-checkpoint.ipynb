```python
#factors of n

n = int(input("Enter the number"))

for i in range(1,n):
    if n%i==0:
        print(i)
```

```python
#add and print todo lists

print("Welcome to our website")

list = []

add = True

while add:
    print("Want to add a task?? Then add as your wish !!!")
    print("----------------------------------------------")
    a = input("Say YES or NO to add task")
    if a=="YES" or a=="yes" or a=="Yes":
        todo = input("Enter ur task here : ")
        list.append(todo)
    else:
        add = False

print("Thank u for using our website")   

print("Your remaining tasks are : ")
for i in range(0,len(list)):
    print(str(i)+")"+" "+list[i])
        
```
