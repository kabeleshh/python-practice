```python
def deg(a,b):
    print(a+" hi... How are You "+b)

deg("hello","fine")    
```

```python
def ghi(a,d):
    print(d+" heyyy..."+a)

ghi(d="hi",a="How are you")    #keyword argument
```

```python
num = 35            # only visible for below codes

def hjk():
    num = 20        # scope of the variable only inside the function
    print(num)

hjk()
print(num)          
```

```python
d = 30

def lkj():
    global d        # declared as global, also used outside the func even visible inside the func
    d = 25
    print(d)

lkj()
print(d)            # o/p : 25
```
